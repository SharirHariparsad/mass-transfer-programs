```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from thermosteam import Chemical,Chemicals, Stream, settings
import biosteam as bst
import CoolProp as CP
```

Setting up thermodynamic properties 

```python
# defiing chemicals 
eth = Chemical('Ethanol')
wat = Chemical('Water')
chems = Chemicals([eth, wat])
settings.set_thermo(chems)

feed = Stream('feed', eth = 70,wat = 30, T=350, P=101325, phase='l')
feed.vle(T=feed, P=feed)
print(f"Feed quality (vapour fraction): {feed.vapor_fraction:.3f}")
print(f"Feed bubble point: {feed.bubble_point_at_P().T - 273.15:.2f} °C")
```

Handeling DIstillations...

```python
# 3. Create the Distillation Column (BinaryDistillation)
# LHK: Light and Heavy Keys (The components being separated)
# y_k and x_k: Desired recovery of the light key in distillate 
# and heavy key in bottoms.
D1 = bst.units.BinaryDistillation('D1', 
                                  ins=feed, 
                                  outs=('distillate', 'bottoms'),
                                  LHK=('Ethanol', 'Water'),
                                  y_k=0.99, # 99% ethanol recovery in top
                                  x_k=0.01, # 1% ethanol recovery in bottom
                                  k=1.2)    # Reflux ratio multiplier (R = k * Rmin)

# 4. Simulate the system
D1.simulate()

# 5. Show results
D1.show()
print(f"Number of theoretical stages: {D1.N_stages}")
```
